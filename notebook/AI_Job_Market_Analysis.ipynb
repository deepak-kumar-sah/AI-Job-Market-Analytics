```python
## Calculate total job openings by country

country_openings = (
    df.groupby("country")["job_openings"]
    .sum()
    .sort_values(ascending=False)
)

# Create bar chart

country_openings.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(country_openings))
)

plt.title("Total Job Openings by Country")
plt.xlabel("Country")
plt.ylabel("Total Job Openings")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../images/total_job_openings_by_country.png",dpi=300,bbox_inches="tight")
plt.show()
```

```python
## Calculate total job openings by job role
role_openings = (
    df.groupby("job_title")["job_openings"]
    .sum()
    .sort_values(ascending=False)
)

role_openings.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(role_openings))
)

plt.title("Total Job Openings by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Total Job Openings")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../images/total_job_openings_by_role.png",dpi=300,bbox_inches="tight")
plt.show()
```

```python
## Calculate average salary by job role

avg_salary = (
    df.groupby("job_title")["salary"]
    .mean()
    .sort_values(ascending=False)
)

avg_salary.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(avg_salary))
    
)

plt.title("Average Salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../images/average_salary_by_job_role.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate Average Salary by Experience Level
avg_salary_experience = (
    df.groupby("experience_level")["salary"]
    .mean()
    .sort_values(ascending=False)
)
avg_salary_experience.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(avg_salary_experience))
)

plt.title("Average Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../images/average_salary_by_experience_level.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate total job openings by experience level

experience_openings = (
    df.groupby("experience_level")["job_openings"]
    .sum()
    .sort_values(ascending=False)
)

experience_openings.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(experience_openings))
)

plt.title("Total Job Openings by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Total job Openings")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../images/total_job_openings_by_experience_level.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate total demand for each skill

skill_demand = (
    df[[
        "skills_python",
        "skills_sql",
        "skills_ml",
        "skills_deep_learning",
        "skills_cloud"
    ]]
    .sum()
    .sort_values(ascending=False)
)

skill_demand.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(skill_demand))
)

plt.title("Skill Demand in AI Job Market")
plt.title("Skills")
plt.ylabel("Number Of job Posting")
plt.xticks(
    ticks=range(len(skill_demand)),
    labels=[
        "Python",
        "SQL",
       " Machine Learning",
       "Deep Learning",
       "Cloud"
    ],
    rotation=45
)
plt.tight_layout()

plt.savefig("../images/skill_demand.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate Average Salary For Each Skill

skill_salary = {
    "Python": df[df["skills_python"]==1]["salary"].mean(),
    "SQL": df[df["skills_sql"]==1]["salary"].mean(),
    "Machine Learning": df[df["skills_ml"]==1]["salary"].mean(),
    "Deep Learning": df[df["skills_deep_learning"]==1]["salary"].mean(),
    "Cloud": df[df["skills_cloud"]==1]["salary"].mean()
}

skill_salary=pd.Series(skill_salary).sort_values(ascending=False)

# create bar chart 
skill_salary.plot(
    kind="bar",
    figsize=(10,5),
    color=sns.color_palette("Set1",len(skill_demand))
)

plt.title("Average Salary by Skill")
plt.xlabel("Skill")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.tight_layout()

# save chart as an image
plt.savefig("../images/average_salary_by_skill.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate number of job posting by year

year_postings = (
    df["job_posting_year"]
    .value_counts()
    .sort_index()
)

# Create line Chart
year_postings.plot(
    kind="line",
    figsize=(10,5),
    marker="o"
)

plt.title("Job Posting by Year")
plt.xlabel("Year")
plt.ylabel("Number of Job Postings")
plt.xticks(year_postings.index)
plt.grid(True)
plt.tight_layout()

# Save chart as an image
plt.savefig("../images/job_postings_by_year.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate Average salary by year
salary_by_year = (
    df.groupby("job_posting_year")["salary"]
    .mean()
    
)

# create line chart
salary_by_year.plot(
    kind="line",
    figsize=(10,5),
    marker="o"
)
plt.title("Average Salary by Year")
plt.xlabel("Year")
plt.ylabel("Average Salary by Year")
plt.xticks(salary_by_year.index)
plt.grid(True)
plt.tight_layout()

# save chart as an image
plt.savefig("../images/average_salary_by_year.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
## Calculate number of job posting by month

month_postings = (
    df["job_posting_month"]
    .value_counts()
    .sort_index()
)

# Create line chart
month_postings.plot(
    kind="line",
    figsize=(10,5),
    marker="o"
)

plt.title("Job Posting by Month")
plt.xlabel("Month")
plt.ylabel("Number of Job Postings")
plt.xticks(range(1,13))
plt.grid(True)
plt.tight_layout()

# save chart as an image
plt.savefig("../images/job_postings_by_month.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
# Calculate average salary by years of experience
experience_salary = (
    df.groupby("years_experience")["salary"]
      .mean()
)

# Create line chart
plt.figure(figsize=(10, 5))

plt.plot(
    experience_salary.index,
    experience_salary.values,
    marker="o"
)

# Add chart labels and title
plt.title("Average Salary by Years of Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Average Salary")

plt.xticks(experience_salary.index)
plt.grid(True, alpha=0.3)

plt.tight_layout()

# Save chart as an image
plt.savefig("../images/average_salary_by_experience_years.png",
            dpi=300,
            bbox_inches="tight")

# Display chart
plt.show()
```

```python
## Total job openings by Company Industry

industry_openings = (
    df.groupby("company_industry")["job_openings"]
    .sum()
    .sort_values(ascending=False)
)
plt.figure(figsize=(10,5))
plt.bar(industry_openings.index,
        industry_openings.values,
        color=sns.color_palette("Set1",len(industry_openings))
)

plt.title("Total Job Openings by Company Industry")
plt.xlabel("Company Industry")
plt.ylabel("Total Job Openings")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../images/total_job_openings_by_company_industry.png",
        dpi=300,
            bbox_inches="tight" )

plt.show()
```

### Key Insights


- Machine Learning Engineer and AI Engineer roles have higher average salaries with the other analyzed job roles.

- Cloud, Machine Learning, SQL, Deep Learning, and Python are among the most frequently demanded skills in the dataset.

- Senior-level positions have the highest average salary, followed by mid-level and entry-level positions.

- High-urgency job postings have a higher average salary than medium- and low-urgency postings.

- Job demand remains relatively consistent across the 2020–2026 period, with yearly job postings showing only moderate variation.
